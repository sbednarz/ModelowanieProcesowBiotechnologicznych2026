## 2. Rola tlenu w bioprocesach

W wielu bioprocesach mikroorganizmy wykorzystują tlen jako końcowy akceptor elektronów w oddychaniu komórkowym.

Schemat uproszczony:

\begin{equation}
\text{substrat} + O_2 \rightarrow CO_2 + H_2O + energia
\end{equation}

Jeżeli transport tlenu jest zbyt wolny:

- tempo wzrostu spada
- metabolizm się zmienia
- pojawiają się produkty uboczne

Dlatego **transport tlenu jest często czynnikiem ograniczającym bioproces**.

---

## 3. OUR – Oxygen Uptake Rate

Szybkość zużycia tlenu przez komórki.

\begin{equation}
OUR = q_{O_2} \cdot X
\end{equation}

gdzie:

- $q_{O_2}$ – specyficzne zużycie tlenu  
- $X$ – stężenie biomasy

Jednostki:

\begin{equation}
OUR \; [\mathrm{mmol\ O_2\ L^{-1}\ h^{-1}}]
\end{equation}


---

## 4. OTR – Oxygen Transfer Rate

Szybkość transportu tlenu z gazu do cieczy.

\begin{equation}
OTR = k_L a (C^* - C)
\end{equation}

gdzie:

- $k_La$ [1/czas] – współczynnik transferu tlenu  
- $C^*$ [mg/L] – stężenie równowagowe tlenu w cieczy
- $C$ [mg/L] – aktualne stężenie tlenu (elektroda DO)


---

# 5. Współczynnik transferu tlenu

\begin{equation}
k_L a = k_L \cdot a
\end{equation}

gdzie:

- $k_L$ – współczynnik przenikania masy  
- $a$ – powierzchnia międzyfazowa


$k_L$, [$m/s$] – współczynnik przenikania

Opisuje szybkość transportu tlenu przez cienką warstwę cieczy przy powierzchni pęcherzyka

Zależy od:
- mieszania (rpm ↑ → $k_L$ ↑)
- turbulencji
- lepkości (lepkość ↑ → $k_L$ ↓)
- temperatury


$a$, [$m^2/m^3$] – powierzchnia międzyfazowa (interfacial area per volume)

To powierzchnia kontaktu gaz–ciecz przypadająca na jednostkę objętości
(powierzchnia pęcherzyków / objętość cieczy).
Zależy od:
- wielkości pęcherzyków (mniejsze → a ↑)
- liczby pęcherzyków
- gas holdup ($ε_g$) - udział objętościowy gazu w całej objętości mieszaniny gaz–ciecz (lub gaz–ciecz–ciało stałe)

Iloczyn $k_L\cdot a$ to całkowita efektywność transferu tlenu - "jak szybko tlen trafia (rozpuszcza się) do medium".

- szybkość mieszanie
- lepsze mieszanie
- większy przepływ gazu
- lepszy dyfuzor
- drobniejsze pęcherzyki
- więcej tlenu w gazie zasilającym


| Typ reaktora / warunki | $k_La$ [h⁻¹] | $k_La$ [s⁻¹] | Uwagi |
|---|---|---|---|
| STR, niskie mieszanie | 50–200 | 0,014–0,056 | lab-scale, wolne mieszadło |
| STR, intensywne napowietrzanie | 200–1000 | 0,056–0,28 | warunki przemysłowe |
| STR, *E. coli* HCDF | 300–1000+ | 0,08–0,28 | przykład modelowy: ~600 h⁻¹ |
| Bubble column (kolumna bąbelkowa) | 100–400 | 0,028–0,11 | zależy od $V_s$ |
| Airlift (pętlowy) | 100–300 | 0,028–0,083 | niższe niż STR przy tej samej mocy |
| Shake flask (kolba) | 10–100 | 0,003–0,028 | często krok ograniczający! |
| Fotobioreaktor (sinice/mikroalgi) | 10–150 | 0,003–0,042 | zależy od geometrii i przepływu CO₂/O₂ |
| Reaktor z grzybnią (*Aspergillus*) | <50 | <0,014 | lepkość grzybni drastycznie obniża $k_La$ |

**Uwaga** Enfors i Shuler używają h⁻¹, Garcia-Ochoa używa s⁻¹.  
Przelicznik: $1\ \text{s}^{-1} = 3600\ \text{h}^{-1}$.

*Źródło: Garcia-Ochoa & Gomez (2009, Biotechnol. Adv. 27:153–176); Shuler & Kargi (2001); Enfors — Fermentation Process Technology*

---

## 6. Warunek stabilnej hodowli

Aby proces był stabilny:

\begin{equation}
OTR \ge OUR
\end{equation}

Jeżeli:

\begin{equation}
OUR > OTR
\end{equation}

występuje **ograniczenie tlenowe**.

---

## 7. Prawo Henry'ego

Rozpuszczalność gazu w cieczy opisuje prawo Henry’ego.

\begin{equation}
C^* = H \cdot p
\end{equation}

gdzie:

- $C^*$ – stężenie gazu w cieczy w stanie równowagi  
- $H$ – stała Henry'ego  
- $p$ – ciśnienie parcjalne (cząstkowe) gazu

Dla tlenu w wodzie (25°C):

\begin{equation}
C^* \approx 8-9\ \mathrm{mg\ L^{-1}}
\end{equation}

---

## 8. Bilans masowy tlenu w bioreaktorze

Zmiana stężenia tlenu w czasie:

\begin{equation}
\frac{dC}{dt} = k_L a (C^* - C) - OUR
\end{equation}

Interpretacja:

- pierwszy człon – transport gaz → ciecz
- drugi człon – zużycie przez komórki



# 9. Gas holdup

Objętościowy udział gazu w cieczy.

\begin{equation}
\varepsilon_g = \frac{V_g}{V}
\end{equation}

gdzie:

- $V_g$ – objętość gazu
- $V$ – objętość układu

---

# 10. Reżimy pracy bioreaktora

Możliwe trzy sytuacje:

1. Nadmiar tlenu

\begin{equation}
OTR \gg OUR
\end{equation}

2. Stan zrównoważony

\begin{equation}
OTR \approx OUR
\end{equation}

3. Ograniczenie tlenowe

\begin{equation}
OUR > OTR
\end{equation}

---
# 11. Typowe zapotrzebowanie na tlen

Zgodnie z Shuler & Kargi (2001), typowe zapotrzebowanie na tlen (OUR — *oxygen uptake rate*) wynosi:

| Organizm | $q_{O_2}$ [mmol O₂ / g·h] | OUR przy X = 10 g/L [mmol/L·h] |
|---|---|---|
| *E. coli* | 5–20 | 50–200 |
| Drożdże (*S. cerevisiae*) | 2–10 | 20–100 |
| Grzyby strzępkowe | 1–5 | 10–50 |
| Sinice / mikroalgi | 0,5–3 | 5–30 |
| Komórki ssacze | 0,05–0,5 | 0,5–5 |

**Konsekwencja praktyczna — przykład HCDF:**

Dla *E. coli* przy $X = 50\ \text{g/L}$ i $q_{O_2} = 10\ \text{mmol/g·h}$:

$$\text{OUR} = q_{O_2} \cdot X = 10 \times 50 = 500\ \text{mmol O}_2\text{/L·h}$$

Przy $C^* \approx 0{,}25\ \text{mmol/L}$ (powietrze, 37°C) i $C_L = 0{,}05\ \text{mmol/L}$:

$$k_La = \frac{\text{OUR}}{C^* - C_L} = \frac{500}{0{,}20} = 2500\ \text{h}^{-1}$$

Tego typowy STR bez wzbogacania w O₂ **nie osiągnie**. Stąd w HCDF konieczne jest:
- wzbogacanie w czysty O₂ (podniesienie $C^*$), lub
- praca pod nadciśnieniem.
